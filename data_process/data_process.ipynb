# Data Processing for Arctic Sea Ice Thickness and Age

## 1. Importing Libraries and Setting Up Environment

```python
import xarray as xr
from utils import build_encoding, generate_monthly_climatology, prepare_ice_age_dataset, convert_to_6931
```

```python
storage_options = {
        "key": "SOME_KEY",
        "secret": "SOME_SECRET",
        "client_kwargs": {
            "endpoint_url": "https://atlantis-vis-o.s3-ext.jc.rl.ac.uk/",
        },
    }
```

```python
ds_cryosat = xr.open_mfdataset(
    "data/cryosat/ice_thickness_nh_ease2-250_*dr-v3p0_*.nc",
    combine="by_coords",
)
ds_cryosat
```

```python
ds_cryosat = convert_to_6931(ds_cryosat)
```

```python
ds_envisat = xr.open_mfdataset(
    "data/envisat/ice_thickness_nh_ease2-250_*dr-v3p0_*.nc",
    combine="by_coords",
)
ds_envisat
```

```python
ds_envisat = convert_to_6931(ds_envisat)
```

```python
ds_arctic = xr.concat(
    [ds_envisat, ds_cryosat],
    dim="time",
    data_vars="minimal",
    coords="minimal",
    compat="override",
    combine_attrs="override",
)

ds_arctic = ds_arctic.sortby("time")
```

```python
ds_arctic
```

```python
ds_arctic = ds_arctic.chunk(
    {
        "time": 1,
        "y": -1,
        "x": -1,
        "nv": -1,
    }
)
```

```python
output_path = "s3://openday/l3_only.zarr"
```

```python
encoding = build_encoding(ds_arctic)
ds_arctic.to_zarr(
    output_path,
    mode="w",
    consolidated=True,
    storage_options=storage_options,
    encoding=encoding,
)
```

```python
ds = xr.open_zarr("https://atlantis-vis-o.s3-ext.jc.rl.ac.uk/openday/l3_only.zarr")
ds
```

```python
ds_ice_age = xr.open_mfdataset(
    "data/ice_age/iceage_nh_12.5km*.nc",
    combine="by_coords",
)
ds_ice_age
```

```python
ds_ice_age = prepare_ice_age_dataset(ds_ice_age)
```

```python
ds_ice_age = generate_monthly_climatology(ds_ice_age)
```

```python
ds_ice_age = ds_ice_age.chunk(
    {
        "month": 1,
        "y": -1,
        "x": -1,
    }
)
```

```python
encoding = build_encoding(ds_ice_age, True)
ds_ice_age["age_of_sea_ice"].attrs["missing_value"] = 21
```

```python
output_path = "s3://openday/age_of_sea_ice_climatology.zarr"
ds_ice_age.to_zarr(
    output_path,
    mode="w",
    consolidated=False,
    storage_options=storage_options,
    encoding=encoding,
    zarr_format=3,
)
```
